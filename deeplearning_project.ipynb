```python
import pandas as pd
import numpy as np
import requests
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from sklearn.metrics.pairwise import cosine_similarity

# Step 1️⃣ - Load Dataset
df = pd.read_csv("travel_dataset.csv")

# Step 2️⃣ - Tokenize Descriptions
tokenizer = Tokenizer()
tokenizer.fit_on_texts(df['Description'])
X_seq = tokenizer.texts_to_sequences(df['Description'])
X_pad = pad_sequences(X_seq, maxlen=25)

# Step 3️⃣ - Build LSTM model (lightweight)
model = Sequential([
    Embedding(input_dim=len(tokenizer.word_index)+1, output_dim=64, input_length=25),
    LSTM(64, return_sequences=False),
    Dense(64, activation='relu')
])
model.compile(optimizer='adam', loss='mse')

# Dummy labels for unsupervised embedding training
y_dummy = np.random.rand(len(X_pad), 64)
model.fit(X_pad, y_dummy, epochs=15, verbose=0)

# Destination embeddings
destination_embeddings = model.predict(X_pad)

# Step 4️⃣ - Cost estimator
def estimate_travel_cost(base_cost, user_country, dest_country):
    if user_country.lower() == dest_country.lower():
        multiplier = 1
        category = "Local travel"
    elif user_country.lower() in ["india", "nepal", "sri lanka", "bangladesh"] and dest_country.lower() in ["india", "nepal", "sri lanka", "bangladesh"]:
        multiplier = 1.5
        category = "Regional travel"
    else:
        multiplier = 3
        category = "International travel"
    return base_cost * multiplier, category

# Step 5️⃣ - USD to INR conversion
def get_usd_to_inr():
    try:
        response = requests.get("https://api.exchangerate-api.com/v4/latest/USD")
        return response.json()["rates"]["INR"]
    except:
        return 84.0

usd_to_inr = get_usd_to_inr()

# Step 6️⃣ - User input
print("🧭 Welcome to Smart Travel Recommender 🌍")
user_country = input("Enter your current country: ").strip()
user_interest = input("Describe your travel interest (e.g., beaches, adventure, temples, nature): ").strip()

# Step 7️⃣ - Encode user interest
seq = tokenizer.texts_to_sequences([user_interest])
pad = pad_sequences(seq, maxlen=25)
user_embedding = model.predict(pad)

# Step 8️⃣ - Calculate similarities
similarities = cosine_similarity(user_embedding, destination_embeddings)[0]
df['Similarity'] = similarities

# Step 9️⃣ - Sort by similarity
df_sorted = df.sort_values(by='Similarity', ascending=False)

# Step 🔟 - Combine: prioritize local first, then global
local_results = df_sorted[df_sorted['Country'].str.lower() == user_country.lower()]
global_results = df_sorted[df_sorted['Country'].str.lower() != user_country.lower()]
top_results = pd.concat([local_results, global_results])

# Step 1️⃣1️⃣ - Display Results
print("\n✅ Top 5 Travel Recommendations for You:")
for i, row in top_results.iterrows():
    dest_country = row['Country']
    base_cost = row['Avg_Cost']
    estimated_cost, travel_type = estimate_travel_cost(base_cost, user_country, dest_country)

    print(f"\n🏙️ City: {row['City']} ({dest_country})")
    print(f"🕒 Best Time: {row['Best_Time_to_Travel']}")
    print(f"📖 Why Visit: {row['Description']}")
    print(f"🌎 Travel Type: {travel_type}")

    # Cost conversion
    if user_country.lower() == "india":
        estimated_cost_inr = estimated_cost * usd_to_inr
        print(f"💰 Estimated Cost: ₹{estimated_cost_inr:,.2f} INR")
    else:
        print(f"💰 Estimated Cost: ${estimated_cost:.2f} USD")

print("\n🎯 (Powered by LSTM for Smart Matching + Local+Global Travel Insights 🌐)")
```
